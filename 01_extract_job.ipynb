```python
url = "https://jobsearch.api.jobtechdev.se/search"

params_all = {
    "offset": 0,
    "limit": 10
}

response_all = requests.get(url, params=params_all)
response_all.raise_for_status()

data_all = response_all.json()

print("Status:", response_all.status_code)
print("Total available jobs:", data_all["total"]["value"])
print("Jobs returned:", len(data_all["hits"]))
```

```python
test_params = {
    "offset": 0,
    "limit": 100
}

test_response = requests.get(url, params=test_params)
test_response.raise_for_status()

test_data = test_response.json()

print("Status:", test_response.status_code)
print("Total available:", test_data["total"]["value"])
print("Requested:", 100)
print("Actually returned:", len(test_data["hits"]))
```

```python
all_jobs = []

batch_size = 100
target_jobs = 500

for offset in range(0, target_jobs, batch_size):

    params = {
        "offset": offset,
        "limit": batch_size
    }

    response = requests.get(url, params=params)
    response.raise_for_status()

    batch_data = response.json()
    jobs_batch = batch_data["hits"]

    all_jobs.extend(jobs_batch)

    print(
        f"Offset: {offset} | "
        f"Retrieved: {len(jobs_batch)} | "
        f"Total collected: {len(all_jobs)}"
    )

print("\nFinal number of jobs collected:", len(all_jobs))
```

```python
# Extract the job IDs from all 500 records
job_ids = [job["id"] for job in all_jobs]

print("Total records:", len(job_ids))
print("Unique Job IDs:", len(set(job_ids)))
print("Duplicate Job IDs:", len(job_ids) - len(set(job_ids)))
```

```python
from datetime import datetime

extract_date = datetime.now().strftime("%Y%m%d_%H%M%S")
raw_filename = f"job_ads_raw_{extract_date}.json"

with open(raw_filename, "w", encoding="utf-8") as f:
    json.dump(all_jobs, f, ensure_ascii=False, indent=2)

print("Raw file saved:", raw_filename)
print("Records saved:", len(all_jobs))
```

```python
df_all = pd.json_normalize(all_jobs)

print("Rows:", df_all.shape[0])
print("Columns:", df_all.shape[1])
```

```python
complex_columns = []

for column in df_all.columns:
    has_complex = df_all[column].apply(
        lambda x: isinstance(x, (list, dict))
    ).any()

    if has_complex:
        complex_columns.append(column)

print("Number of columns containing lists/dictionaries:", len(complex_columns))

for column in complex_columns:
    print(column)
```

```python
skills_example = df_all[
    df_all["must_have.skills"].apply(
        lambda x: isinstance(x, list) and len(x) > 0
    )
][["id", "headline", "must_have.skills"]]

print("Jobs with must-have skills:", len(skills_example))

skills_example.head(5)
```

```python
# Remove columns containing lists/dictionaries from the main staging dataset
staging_columns = [
    col for col in df_all.columns
    if col not in complex_columns
]

df_staging = df_all[staging_columns].copy()

print("Original columns:", df_all.shape[1])
print("Complex columns excluded:", len(complex_columns))
print("Staging columns:", df_staging.shape[1])
print("Staging rows:", df_staging.shape[0])
```

```python
job_skills = []

for job in all_jobs:
    job_id = job["id"]

    for skill in job["must_have"]["skills"]:
        job_skills.append({
            "job_id": job_id,
            "skill_concept_id": skill.get("concept_id"),
            "skill_label": skill.get("label"),
            "skill_weight": skill.get("weight")
        })

df_job_skills = pd.DataFrame(job_skills)

print("Skill requirement rows:", len(df_job_skills))
print("Jobs represented:", df_job_skills["job_id"].nunique())

df_job_skills.head(10)
```

```python
requirement_columns = [
    "must_have.skills",
    "must_have.languages",
    "must_have.work_experiences",
    "must_have.education",
    "must_have.education_level",
    "nice_to_have.skills",
    "nice_to_have.languages",
    "nice_to_have.work_experiences",
    "nice_to_have.education",
    "nice_to_have.education_level"
]

for column in requirement_columns:
    jobs_with_values = df_all[column].apply(
        lambda x: isinstance(x, list) and len(x) > 0
    ).sum()

    print(f"{column}: {jobs_with_values} jobs")
```

```python
fields_to_inspect = [
    "must_have.languages",
    "must_have.work_experiences",
    "must_have.education"
]

for column in fields_to_inspect:
    example = df_all[
        df_all[column].apply(
            lambda x: isinstance(x, list) and len(x) > 0
        )
    ][["id", "headline", column]].iloc[0]

    print("\nFIELD:", column)
    print("Job ID:", example["id"])
    print("Headline:", example["headline"])
    print("Value:", example[column])
```

```python
example = df_all[
    df_all["must_have.education_level"].apply(
        lambda x: isinstance(x, list) and len(x) > 0
    )
][["id", "headline", "must_have.education_level"]].iloc[0]

print("Job ID:", example["id"])
print("Headline:", example["headline"])
print("Education Level:", example["must_have.education_level"])
```

```python
requirement_mapping = {
    "must_have": {
        "skills": "Skill",
        "languages": "Language",
        "work_experiences": "Work Experience",
        "education": "Education",
        "education_level": "Education Level"
    },
    "nice_to_have": {
        "skills": "Skill",
        "languages": "Language",
        "work_experiences": "Work Experience",
        "education": "Education",
        "education_level": "Education Level"
    }
}

requirements = []

for job in all_jobs:
    job_id = job["id"]

    for level, types in requirement_mapping.items():

        for source_field, requirement_type in types.items():

            values = job.get(level, {}).get(source_field, [])

            for value in values:
                requirements.append({
                    "job_id": job_id,
                    "concept_id": value.get("concept_id"),
                    "requirement_label": value.get("label"),
                    "requirement_type": requirement_type,
                    "requirement_level": (
                        "Must Have"
                        if level == "must_have"
                        else "Nice to Have"
                    ),
                    "weight": value.get("weight"),
                    "legacy_ams_taxonomy_id":
                        value.get("legacy_ams_taxonomy_id")
                })

df_requirements = pd.DataFrame(requirements)

print("Requirement rows:", len(df_requirements))
print("Jobs with structured requirements:",
      df_requirements["job_id"].nunique())

df_requirements.head(10)
```

```python
duplicate_requirements = df_requirements.duplicated(
    subset=[
        "job_id",
        "concept_id",
        "requirement_type",
        "requirement_level"
    ]
).sum()

print("Total requirement rows:", len(df_requirements))
print("Duplicate requirement rows:", duplicate_requirements)
print(
    "Unique requirement combinations:",
    len(df_requirements) - duplicate_requirements
)
```

```python
# Save the main job-ad data
df_staging.to_csv(
    "job_ads_source.csv",
    index=False,
    encoding="utf-8-sig"
)

# Save the requirements data
df_requirements.to_csv(
    "job_requirements_source.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Files saved successfully.")
print("Job ads:", len(df_staging))
print("Requirements:", len(df_requirements))
```
